# Labeling Time Analysis

How much time does a volunteer need to label a single image? This is an important question to answer in order to estimate the time needed to label a large dataset. In this notebook, we will analyze the time taken by volunteers to label images in the dataset.

```python
import pandas as pd

from zooniverse.config import get_config
from pathlib import Path

phase_tag = "Iguanas 4th launch"

output_path = Path(f"/Users/christian/data/zooniverse/2024_09_27_analysis").joinpath(phase_tag)
output_path.mkdir(parents=True, exist_ok=True)
input_path = Path("/Users/christian/data/zooniverse")
config = get_config(phase_tag=phase_tag, input_path=input_path, output_path=output_path)

df_yes_no = pd.read_csv(config["yes_no_dataset"])
```

```python
df_yes_no
```

```python
df_yes_no["labeling_started_at"] = pd.to_datetime(df_yes_no["labeling_started_at"])
df_yes_no["labeling_finished_at"] = pd.to_datetime(df_yes_no["labeling_finished_at"])

df_yes_no["time_spent"] =  df_yes_no['labeling_finished_at'] - df_yes_no['labeling_started_at']
df_yes_no["time_spent"]
```

```python
df_yes_no
```

```python
df_yes_no.groupby("subject_id")["time_spent"].sum()
```

```python
df_yes_no.groupby("subject_id")["time_spent"].sum().describe()
```

```python
df_yes_no.groupby("subject_id")["time_spent"].sum().median()
```
